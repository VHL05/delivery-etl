```python
import pandas as pd
```

```python
# 1. Đọc dữ liệu

print("Reading data from CSV file: olist_orders_dataset.csv ...")
df_orders = pd.read_csv('../data/raw/olist_orders_dataset.csv')
```

```python
# 2. Kiểm tra tổng quan và null

print(f"Tổng số dòng: {len(df_orders)}")
print("Tỉ lệ null (%) ở các cột quan trọng: ")
null_rates = df_orders[['order_approved_at', 'order_delivered_carrier_date', 'order_delivered_customer_date']].isnull().mean() * 100
print(null_rates)
```

```python
# 3. Ép kiểu dữ liệu từ string qua datetime

time_cols = [
    'order_purchase_timestamp',
    'order_approved_at',
    'order_delivered_carrier_date',
    'order_delivered_customer_date',
    'order_estimated_delivery_date'
]
print("Converting string to datetime for columns: ", time_cols)
for col in time_cols:
    df_orders[col] = pd.to_datetime(df_orders[col])
```

```python
# 4. Xác nhận order_id là duy nhất
is_unique = df_orders['order_id'].is_unique
if is_unique:
    print("order_id is unique.")
else:
    print("order_id is NOT unique. There are duplicates.")
```

```python
# 5. Xuất file clean 
output_path = '../data/processed_local/orders_clean.csv'
df_orders.to_csv(output_path, index=False)
print(f"Cleaned data exported to: {output_path}")
```
